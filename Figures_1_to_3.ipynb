# Radial-flow fluctuations: Figures 1–3

Replot the model/data comparisons from [L. Du, Phys. Rev. C **113**, 014901 (2026)](https://arxiv.org/abs/2508.07184v2) using the supplied numerical inputs. The default model results for Figures 2–3 are the original files in [Zenodo 17808246](https://zenodo.org/records/17808246).

Run this notebook from the collection's root after installing `requirements.txt`. It reads only the bundled files and produces new plots; no simulation or particle-level analysis is required. See [README](README.md#data-and-conventions) for source citations, selections and uncertainty conventions.

```python
%matplotlib inline
from pathlib import Path
from IPython.display import display
import matplotlib.pyplot as plt
from radial_flow import load_inputs, plot_figure1, plot_figure2, plot_figure3

inputs = load_inputs()
output = Path('figures')
output.mkdir(exist_ok=True)
```

## Figure 1: spectra and elliptic flow

Au+Au at 200 GeV: $\pi^+$, $K^+$ and proton invariant spectra in 30–40%, and charged-hadron $v_2(p_T)$ in 0–5% and 30–40%. Spectra are multiplied by 100, 10 and 1 for display only. Markers are PHENIX spectra and STAR flow measurements; lines are the default calibrated model. No model uncertainty for these curves is supplied.

```python
fig = plot_figure1(inputs)
fig.savefig(output / 'figure1.png')
display(fig)
plt.close(fig)
```

## Figure 2: charged-hadron radial-flow fluctuations

Pb+Pb at 5.02 TeV, 30–40%. Solid lines use the ALICE-like or ATLAS-like momentum selections; dashed lines include particles down to $p_T=0$. Shading is the saved model statistical uncertainty. ALICE and ATLAS markers use official HEPData measurements. Experimental error bars combine the statistical and systematic components in quadrature; both components are retained separately in the CSV.

```python
fig = plot_figure2(inputs)
fig.savefig(output / 'figure2.png')
display(fig)
plt.close(fig)
```

## Figure 3: identified-hadron radial-flow fluctuations

Pb+Pb at 5.02 TeV, 30–40%. Compare the saved fluctuation form (solid) and covariance form (dashed) for $\pi^\pm$, $K^\pm$ and $p/\bar p$. Shading is the saved statistical uncertainty; ALICE markers and experimental errors come from the official HEPData tables. Error bars combine statistical and systematic components in quadrature. The two methods and the subevent convention are described in Sec. III A and Appendix A of the paper.

```python
fig = plot_figure3(inputs)
fig.savefig(output / 'figure3.png')
display(fig)
plt.close(fig)
```

## Reuse the functions

`read_zenodo(path)` reads other result files from the same deposit. `plot_figure1(inputs)`, `plot_figure2(inputs)` and `plot_figure3(inputs)` each return a Matplotlib figure, so their labels, axes and output format can be adjusted. For example, `fig.savefig('my_figure.svg')` exports a vector image. These examples reuse saved results and do not recompute observables from particle events.
